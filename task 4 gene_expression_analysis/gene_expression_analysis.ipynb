# Gene Expression Analysis

*This notebook analyzes an ACE-related gene expression dataset from NCBI GEO (GSE199578) and identifies the top 10 expressed genes based on average expression across all samples.*

```python
import pandas as pd

# Read the compressed gene expression dataset
df = pd.read_csv(
    "GSE199578_featureCounts_results.txt.gz",
    sep="\t",
    compression="gzip",
    skiprows=1
)

# Select gene expression sample columns
sample_columns = df.columns[6:]

# Calculate average expression for each gene
df["Average_Expression"] = df[sample_columns].mean(axis=1)

# Sort genes by average expression and select the top 10
top_10_genes = (
    df.sort_values(
        by="Average_Expression",
        ascending=False
    )[["Geneid", "Average_Expression"]]
    .head(10)
)

# Display the top 10 expressed genes
print("Top 10 Expressed Genes:")
print(top_10_genes)

# Save the results
top_10_genes.to_csv(
    "top_10_expressed_genes.csv",
    index=False
)
```

## Results

The analysis calculated the average gene expression across the eight samples and ranked all genes from highest to lowest expression.

The top 10 expressed genes were identified as ACTB, ACTG1, STMN1, ENO1, MT-CO1, GAPDH, PRCP, ATP5F1B, LDHA, and FTL.

Among these genes, ACTB showed the highest average expression, while FTL ranked tenth. These results represent the top 10 genes based on their average raw expression counts across all samples in the dataset.
